```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data    
```

```python
data = load_data("data.json")
```

```python
def display(data):
    print("information about the users")
    for user in data['users']:
        print(f"{user['name']} name of the user and the friends of the user are as follows {user['friends']}")
    print("page id and page name")
    for page in data['pages']:
         print(f" ID: {page['id']} name: {page['name']}")
     
```
