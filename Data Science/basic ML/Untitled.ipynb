```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [[150,0],[160,0],[110,1],[120,1]]
```

```python
labels=["apple","apple","orange","orange"]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features,labels)
```

```python
clf.predict([[170,1]])
```

```python
clf.predict([[200,1]])
```
